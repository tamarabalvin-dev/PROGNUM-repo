```python
#4.1
x = [1,2,3,10]
y = [10,20,30,40]

#function zip()
z = list(zip(x,y))
for i in range(len(z)):
    z[i] = z[i][0] + z[i][1]
print(z)

#one line
z = [a+b for a,b in zip(x,y)]
print(z)

#function map()
def combine(list1,list2):
    combined = []
    for i in range(len(list1)):
        addd = list1[i] + list2[i]
        combined += [addd]
    return(combined)

print(combine(x,y))
```

```python
#4.3

import numpy as np

throws = ['R', 'P', 'S'] #possible throws
user = input('Welcome to Rock-Paper-Scissors! To play, enter R, P, or S: ') #user throw
program = np.random.choice(throws) #program throw

#idiot proofing
if not user == 'R' and not user == 'S' and not user == 'P':
    print(f'You did not input a valid letter.')
    
#actual game
else:
    print(f'You played: {user}. The computer plays: {program}.')
    if user == program:
        print(f'No one won!') #tie
    elif (user == 'R' and program == 'S') or (user == 'P' and program == 'R') or (user == 'S' and program == 'P'):
        print(f'You won!') #checks all win conditions at once
    else:
        print(f'You lost!') #only possibility left is loss
```

```python
#4.4

import numpy as np, matplotlib.pyplot as plt, scipy

dis = np.random.normal(185,8,100000)

mean = np.mean(dis)
std = np.std(dis)

h = 190
cdf = scipy.stats.norm.cdf(h,8,185)
P = 100*cdf
print(f'Probability of values smaller than {h}: {P:.2f}%')

cdf1 = scipy.stats.norm.cdf(mean-std,scale=8,loc=185)
P1 = 100*cdf1
cdf2 = scipy.stats.norm.cdf(mean+std,scale=8,loc=185)
P2 = 100*cdf2
print(f'Probability of values within 1 standard deviation: {(P2-P1):.2f}%')

plt.hist(dis, bins=40,label='Normal distribution')
plt.axvline(mean,color='green',label=fr'Mean = {mean:.2f}')
plt.axvspan(mean-std,mean+std,alpha=0.2,color='red',label=fr'Standard deviation = {std:.2f}')
plt.legend()
plt.show()
```

```python
#4.16

from astropy.cosmology import Planck18
import scipy

H0 = Planck18.H0.value
Om0 = Planck18.Om0
Ode0 = Planck18.Ode0

z = 2

def tL(z):
    Hz = (H0*3.24e-20)*(Om0*((1+z)**3)+Ode0)**(1/2)
    tL = 1/((1+z)*Hz)
    return tL

result = (scipy.integrate.quad(tL,0,z)[0])/(3.15e16)
print(f'The lookback time in Gyr: {result}')
```

```python
#4.21

import matplotlib.pyplot as plt, numpy as np, scipy

x = np.random.normal(0,1,100000)
y = np.random.normal(2,1,100000)

plt.hist(x,bins=30,alpha=0.5,color='tomato',density=True,label='X',lw=1,edgecolor='black')
plt.hist(y,bins=30,alpha=0.5,color='seagreen',density=True,label='Y',lw=1,edgecolor='black')

xsort = np.linspace(x.min(),x.max(),100000)
ysort = np.linspace(y.min(),y.max(),100000)

xpdf = scipy.stats.norm.pdf(xsort,0,1)
ypdf = scipy.stats.norm.pdf(ysort,2,1)

plt.plot(xsort,xpdf,color='black',linestyle='dotted',lw=4,label='X(PDF)')
plt.plot(ysort,ypdf,color='black',linestyle='dashed',lw=4,label='Y(PDF)')
plt.legend()
plt.title('1D Histogram + PDF')
plt.xlabel('Values')
plt.ylabel('Probability Density')
plt.grid(axis='y',linestyle='dashed')
plt.text(-5,0.15,'$\mathcal{N}$(0,1)',color='tomato',fontsize=15)
plt.text(5,0.15,'$\mathcal{N}$(2,1)',color='seagreen',fontsize=15)
plt.arrow(-3,0.16,0.5,0,width=0.02,head_width=0.04,head_length=0.5,color='tomato')
plt.arrow(5,0.16,-0.5,0,width=0.02,head_width=0.04,head_length=0.5,color='seagreen')
plt.xticks(np.arange(-6,9,step=1))
plt.show()
```

```python
import matplotlib.pyplot as plt, numpy as np, scipy

x = np.random.normal(0,1,10000)
y = np.random.normal(2,1,10000)

counts, xedges, yedges = np.histogram2d(x, y, bins=30)

sigmas = [1., 2., 3.]
levels = []
fmt = {}

for i in range(len(sigmas)):
    levels.append(float(sigmas[i]) * np.std(counts))
    fmt[float(sigmas[i]) * np.std(counts)] = r"${}\sigma$".format(int(sigmas[-(i+1)]))

plt.hist2d(x,y,bins=30,cmap='Blues')
plt.colorbar()
plt.title('2D Histogram + Contour')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(linestyle='dashed',color='black')
cs = plt.contour(xedges[:-1], yedges[:-1], counts.T, levels, colors='red')
plt.clabel(cs, fontsize=12, fmt=fmt)
plt.show()
```
