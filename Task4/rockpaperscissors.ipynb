```python
import numpy as np

throws = ['R', 'P', 'S'] #possible throws
user = input('Welcome to Rock-Paper-Scissors! To play, enter R, P, or S: ') #user throw
program = np.random.choice(throws) #program throw

#idiot proofing
if not user == 'R' and not user == 'S' and not user == 'P':
    print(f'You did not input a valid letter.')
    
#actual game
else:
    print(f'You played: {user}. The computer plays: {program}.')
    if user == program:
        print(f'No one won!') #tie
    elif (user == 'R' and program == 'S') or (user == 'P' and program == 'R') or (user == 'S' and program == 'P'):
        print(f'You won!') #checks all win conditions at once
    else:
        print(f'You lost!') #only possibility left is loss
```
