```python
#4.1
x = [1,2,3,10]
y = [10,20,30,40]

#function zip()
z = list(zip(x,y))
for i in range(len(z)):
    z[i] = z[i][0] + z[i][1]
print(z)

#one line
z = [a+b for a,b in zip(x,y)]
print(z)

#function map()
def combine(list1,list2):
    combined = []
    for i in range(len(list1)):
        addd = list1[i] + list2[i]
        combined += [addd]
    return(combined)

print(combine(x,y))
```

```python
#4.3

import numpy as np

throws = ['R', 'P', 'S'] #possible throws
user = input('Welcome to Rock-Paper-Scissors! To play, enter R, P, or S: ') #user throw
program = np.random.choice(throws) #program throw

#idiot proofing
if not user == 'R' and not user == 'S' and not user == 'P':
    print(f'You did not input a valid letter.')
    
#actual game
else:
    print(f'You played: {user}. The computer plays: {program}.')
    if user == program:
        print(f'No one won!') #tie
    elif (user == 'R' and program == 'S') or (user == 'P' and program == 'R') or (user == 'S' and program == 'P'):
        print(f'You won!') #checks all win conditions at once
    else:
        print(f'You lost!') #only possibility left is loss
```

```python
#4.4

import numpy as np, matplotlib.pyplot as plt, scipy

dis = np.random.normal(185,8,100000)

mean = np.mean(dis)
std = np.std(dis)

h = 190
cdf = scipy.stats.norm.cdf(h,8,185)
P = 100*cdf
print(f'Probability of values smaller than {h}: {P:.2f}%')

cdf1 = scipy.stats.norm.cdf(mean-std,scale=8,loc=185)
P1 = 100*cdf1
cdf2 = scipy.stats.norm.cdf(mean+std,scale=8,loc=185)
P2 = 100*cdf2
print(f'Probability of values within 1 standard deviation: {(P2-P1):.2f}%')

plt.hist(dis, bins=40,label='Normal distribution')
plt.axvline(mean,color='green',label=fr'Mean = {mean:.2f}')
plt.axvspan(mean-std,mean+std,alpha=0.2,color='red',label=fr'Standard deviation = {std:.2f}')
plt.legend()
plt.show()
```
